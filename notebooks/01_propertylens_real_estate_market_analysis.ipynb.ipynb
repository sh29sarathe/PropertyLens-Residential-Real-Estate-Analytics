```python
from pathlib import Path
import pandas as pd
import numpy as np

# Find the project folder

current_folder = Path.cwd()

if current_folder.name == "notebooks":
    project_folder = current_folder.parent
else:
    project_folder = current_folder

data_folder = project_folder / "data"


# Find CSV files

csv_files = list(data_folder.glob("*.csv"))

if len(csv_files) == 0:
    raise FileNotFoundError(
        f"No CSV file was found in: {data_folder}\n\n"
        "Please put your real estate CSV file inside the data folder and run this cell again."
    )

if len(csv_files) > 1:
    print("More than one CSV file was found:")
    for file in csv_files:
        print("-", file.name)

    raise ValueError(
        "\nPlease keep only the dataset we are using for this project inside the data folder."
    )

csv_file = csv_files[0]

print("Dataset found:", csv_file.name)
print("")


# Load the dataset


df = pd.read_csv(csv_file)

print("Dataset loaded successfully.")
print("")


# Basic dataset information

print("BASIC DATASET INFORMATION")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nColumn names:")
for column in df.columns:
    print("-", column)


# Data types

print("DATA TYPES")

data_types = pd.DataFrame({
    "column": df.columns,
    "data_type": df.dtypes.astype(str).values
})

print(data_types.to_string(index=False))


# Missing values

print("MISSING VALUES")

missing_values = pd.DataFrame({
    "column": df.columns,
    "missing_count": df.isna().sum().values,
    "missing_percent": (
        df.isna().mean().mul(100).round(2).values
    )
})

missing_values = missing_values.sort_values(
    by="missing_count",
    ascending=False
)

print(missing_values.to_string(index=False))


# Duplicate rows

print("DUPLICATE ROWS")

duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)


# Number of unique values

print("UNIQUE VALUES BY COLUMN")

unique_values = pd.DataFrame({
    "column": df.columns,
    "unique_count": [df[column].nunique(dropna=False) for column in df.columns]
})

print(unique_values.to_string(index=False))


# Categorical columns

print("CATEGORICAL COLUMN VALUES")

categorical_columns = df.select_dtypes(
    include=["object", "string"]
).columns.tolist()

if len(categorical_columns) == 0:
    print("No object, category, or boolean columns were found.")
else:
    for column in categorical_columns:
        print(f"\n{column}")
        print("-" * len(column))

        values = df[column].value_counts(
            dropna=False
        )

        print(values.to_string())


# Numerical columns

print("NUMERICAL COLUMN SUMMARY")

numerical_columns = df.select_dtypes(
    include=np.number
).columns.tolist()

if len(numerical_columns) == 0:
    print("No numerical columns were found.")
else:
    numerical_summary = df[numerical_columns].describe().T

    numerical_summary["missing_count"] = df[numerical_columns].isna().sum()
    numerical_summary["zero_count"] = (
        df[numerical_columns] == 0
    ).sum()
    numerical_summary["negative_count"] = (
        df[numerical_columns] < 0
    ).sum()

    print(
        numerical_summary[
            [
                "count",
                "mean",
                "std",
                "min",
                "25%",
                "50%",
                "75%",
                "max",
                "missing_count",
                "zero_count",
                "negative_count"
            ]
        ].round(2).to_string()
    )


# First few rows

print("FIRST 5 ROWS")

print(df.head().to_string())


# Last few rows

print("LAST 5 ROWS")

print(df.tail().to_string())


# Possible text problems

print("POSSIBLE TEXT / CATEGORY ISSUES")

for column in categorical_columns:
    values = df[column].dropna().astype(str)

    leading_space_count = values.str.startswith(" ").sum()
    trailing_space_count = values.str.endswith(" ").sum()

    if leading_space_count > 0 or trailing_space_count > 0:
        print(
            f"{column}: "
            f"{leading_space_count} values with leading spaces, "
            f"{trailing_space_count} values with trailing spaces"
        )


# Possible numeric columns

print("COLUMNS CURRENTLY READ AS TEXT")

text_columns = df.select_dtypes(include=["object", "string"]).columns.tolist()

if len(text_columns) == 0:
    print("No text columns were found.")
else:
    print(
        "These columns are currently stored as text. "
        "Some may contain numbers that need checking later:"
    )

    for column in text_columns:
        print("-", column)


print(
    "No cleaning or changes have been made to the dataset."
)
```
