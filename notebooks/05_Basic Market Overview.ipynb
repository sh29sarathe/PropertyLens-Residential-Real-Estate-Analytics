```python
import pandas as pd
from pathlib import Path

project_folder = Path.cwd()

if project_folder.name == "notebooks":
    project_folder = project_folder.parent

data_folder = project_folder / "data"
feature_file = data_folder / "propertylens_featured.csv"

df = pd.read_csv(feature_file)

# Overall market summary
market_summary = pd.DataFrame({
    "Metric": [
        "Total properties",
        "Average price",
        "Median price",
        "Minimum price",
        "Maximum price",
        "Average area",
        "Median area",
        "Average rate per sqft",
        "Median rate per sqft"
    ],
    "Value": [
        len(df),
        df["price"].mean(),
        df["price"].median(),
        df["price"].min(),
        df["price"].max(),
        df["area"].mean(),
        df["area"].median(),
        df["rate_per_sqft"].mean(),
        df["rate_per_sqft"].median()
    ]
})

print(market_summary.to_string(index=False))


# Property type distribution
print("\nFlat type:")
print(df["flat_type"].value_counts())


# Status distribution
print("\nStatus:")
print(df["status"].value_counts())


# RERA distribution
print("\nRERA approval:")
print(df["rera_approval"].value_counts())


# BHK distribution
print("\nBHK:")
print(df["bhk_group"].value_counts())


# Localities with the most listings
print("\nTop 15 localities by number of listings:")
print(df["locality"].value_counts().head(15))


# Price summary by flat type
print("\nPrice summary by flat type:")
print(
    df.groupby("flat_type")["price_lakh"]
    .agg(["count", "mean", "median", "min", "max"])
    .sort_values("median", ascending=False)
)


# Rate summary by flat type
print("\nRate per sqft summary by flat type:")
print(
    df.groupby("flat_type")["rate_per_sqft"]
    .agg(["count", "mean", "median", "min", "max"])
    .sort_values("median", ascending=False)
)
```
