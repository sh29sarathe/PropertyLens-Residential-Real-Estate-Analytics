```python
import pandas as pd
from pathlib import Path

project_folder = Path.cwd()

if project_folder.name == "notebooks":
    project_folder = project_folder.parent

data_folder = project_folder / "data"
feature_file = data_folder / "propertylens_featured.csv"

df = pd.read_csv(feature_file)

# BHK-level analysis
bhk_summary = (
    df.groupby("bhk_group")
    .agg(
        listings=("bhk_group", "size"),
        avg_price_lakh=("price_lakh", "mean"),
        median_price_lakh=("price_lakh", "median"),
        avg_area_sqft=("area_sqft", "mean"),
        median_area_sqft=("area_sqft", "median"),
        avg_rate_per_sqft=("rate_per_sqft", "mean"),
        median_rate_per_sqft=("rate_per_sqft", "median")
    )
    .reset_index()
)

print("BHK analysis:")
print(
    bhk_summary
    .sort_values("listings", ascending=False)
    .to_string(index=False)
)


# Property type analysis
property_type_summary = (
    df.groupby("flat_type")
    .agg(
        listings=("flat_type", "size"),
        avg_price_lakh=("price_lakh", "mean"),
        median_price_lakh=("price_lakh", "median"),
        avg_area_sqft=("area_sqft", "mean"),
        median_area_sqft=("area_sqft", "median"),
        avg_rate_per_sqft=("rate_per_sqft", "mean"),
        median_rate_per_sqft=("rate_per_sqft", "median")
    )
    .reset_index()
)

print("\nProperty type analysis:")
print(
    property_type_summary
    .sort_values("listings", ascending=False)
    .to_string(index=False)
)


# Area versus price by area band
area_band_summary = (
    df.groupby("area_band")
    .agg(
        listings=("area_band", "size"),
        avg_area_sqft=("area_sqft", "mean"),
        median_area_sqft=("area_sqft", "median"),
        avg_price_lakh=("price_lakh", "mean"),
        median_price_lakh=("price_lakh", "median"),
        avg_rate_per_sqft=("rate_per_sqft", "mean"),
        median_rate_per_sqft=("rate_per_sqft", "median")
    )
    .reset_index()
)

print("\nArea band analysis:")
print(
    area_band_summary
    .to_string(index=False)
)


# Relationship between BHK and property size
bhk_area_summary = (
    df[df["bhk_group"] != "Not specified"]
    .groupby("bhk_group")
    .agg(
        listings=("bhk_group", "size"),
        median_area_sqft=("area_sqft", "median"),
        median_price_lakh=("price_lakh", "median"),
        median_rate_per_sqft=("rate_per_sqft", "median")
    )
    .reset_index()
)

print("\nBHK versus area and price:")
print(
    bhk_area_summary
    .sort_values("bhk_group")
    .to_string(index=False)
)


# Relationship between BHK and property type
bhk_property_type = pd.crosstab(
    df["bhk_group"],
    df["flat_type"]
)

print("\nBHK and property type:")
print(bhk_property_type.to_string())
```
