```python
import pandas as pd
from pathlib import Path

project_folder = Path.cwd()

if project_folder.name == "notebooks":
    project_folder = project_folder.parent

data_folder = project_folder / "data"
feature_file = data_folder / "propertylens_featured.csv"

df = pd.read_csv(feature_file)

# BHK values that need investigation
bhk_check = df[df["bhk_count"].isin([0]) | (df["bhk_count"] >= 8)].copy()

print("BHK values needing investigation:")
print(
    bhk_check[
        [
            "bhk_count",
            "area",
            "price",
            "rate_per_sqft",
            "property_type",
            "locality",
            "flat_type"
        ]
    ].sort_values("bhk_count").to_string(index=False)
)

print("\nBHK counts:")
print(bhk_check["bhk_count"].value_counts().sort_index())


# Very large area values
area_check = df[df["area"] >= 10000].copy()

print("\nProperties with area >= 10,000 sqft:")
print(
    area_check[
        [
            "area",
            "price",
            "rate_per_sqft",
            "property_type",
            "locality",
            "flat_type"
        ]
    ].sort_values("area", ascending=False).to_string(index=False)
)


# Very high price values
price_check = df[df["price_crore"] >= 20].copy()

print("\nProperties priced at 20 crore or more:")
print(
    price_check[
        [
            "price",
            "price_crore",
            "area",
            "rate_per_sqft",
            "property_type",
            "locality",
            "flat_type"
        ]
    ].sort_values("price", ascending=False).to_string(index=False)
)


# Very high rate per sqft values
rate_check = df[df["rate_per_sqft"] >= 50000].copy()

print("\nProperties with rate per sqft >= 50,000:")
print(
    rate_check[
        [
            "rate_per_sqft",
            "price",
            "price_crore",
            "area",
            "property_type",
            "locality",
            "flat_type"
        ]
    ].sort_values("rate_per_sqft", ascending=False).to_string(index=False)
)


# Check whether extreme area values are mainly plots
print("\nFlat type for properties with area >= 10,000 sqft:")
print(area_check["flat_type"].value_counts())


# Check whether zero BHK values are mainly plots
zero_bhk = df[df["bhk_count"] == 0]

print("\nFlat type for BHK = 0:")
print(zero_bhk["flat_type"].value_counts())

print("\nProperty type for BHK = 0:")
print(zero_bhk["property_type"].value_counts().head(20))
```
